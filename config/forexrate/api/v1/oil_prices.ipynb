BRENT (EUR and GBP)

```python
from fredapi import Fred
import pandas as pd 
from dotenv import dotenv_values
config = dotenv_values()
api_key=config.get("FRED_API_KEY")

fred = Fred(api_key)
print(fred)

oil_rate = fred.get_series('DCOILBRENTEU', start_date='2014-01-01', end_date='2024-12-20')  


eur = pd.DataFrame({
    'EUR Rate': oil_rate,
})

eur.to_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/oil rates/brent.csv", index=True)


print(eur.head(5))
```
