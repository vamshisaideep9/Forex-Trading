```python
import pandas as pd 

file_paths = {
    'CAD' : "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/Trade balance/cad_trade_balance.csv",
    "CNY" : "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/Trade balance/cny_trade_balance.csv",
    "EUR" : "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/Trade balance/euro_trade_balance.csv",
    "GBP" : "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/Trade balance/gbp_trade_balance.csv",
    "JPY" : "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/Trade balance/JPY_trade_balance.csv",
    "USA" : "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/Trade balance/us_trade_balance.csv"
}


data_frames = {}
for country, file_path in file_paths.items():
    df = pd.read_csv(file_path)
    df['Country'] = country

    if 'Trade Balance (M)' in df.columns:
        df.rename(columns={'Trade Balance (M)': 'Trade_Balance'}, inplace=True)
        df['Trade_Balance'] /= 1000  
    elif 'Trade Balance (B)' in df.columns:
        df.rename(columns={'Trade Balance (B)': 'Trade_Balance'}, inplace=True)

    df['date'] = pd.to_datetime(df['date'])
    data_frames[country] = df[['date', 'Trade_Balance']].rename(columns={'Trade_Balance': country})

result = None
for country, df in data_frames.items():
    if result is None:
        result = df
    else:
        result = pd.merge(result, df, on='date', how='outer')


result.sort_values(by='date', inplace=True)

result.index = result.index + 1
result.to_csv('C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/Trade balance/cleaned_data_trade_balance.csv', index=True)

```
