Data cleaning of WTI

```python
wti = pd.read_csv("c:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/oil rates/wti.csv")
wti.head(10)
```

```python
wti.isna().value_counts()
```

```python
wti['USD CAD oil prices(dollars)'] = wti['USD CAD oil prices(dollars)'].interpolate()
wti.head(50)
```

```python
wti.isna().value_counts()
```

```python
wti = wti.fillna(wti['USD CAD oil prices(dollars)'].mean())
```

```python
wti.index = wti.index + 1
wti.head(10)
```

```python
wti.to_csv("c:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/forexrate/data/oil rates/updated_wti.csv")
```
