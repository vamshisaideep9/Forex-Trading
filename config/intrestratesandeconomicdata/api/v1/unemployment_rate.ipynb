USA UNEMPLOYMENT RATE

```python
from fredapi import Fred
import pandas as pd 
from dotenv import dotenv_values
config = dotenv_values()
api_key=config.get("FRED_API_KEY")

fred = Fred(api_key)


usa_rate = fred.get_series('UNRATE', start_date='2014-01-01', end_date='2024-12-20')


usa_rate_df = pd.DataFrame(usa_rate, columns=["USA Unemployment Rate"])


usa_rate_df = usa_rate_df.resample('D').ffill()

usa_rate_df.reset_index(inplace=True)

usa_rate_df.index = usa_rate_df.index + 1

usa_rate_df.columns = ['Date', 'USA Unemployment Rate']


print(usa_rate_df.head(10))

usa_rate_df.to_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/Unemployment rate/usa_unemployment_rate.csv")
```
