```python
import pandas as pd

# Define file paths
usd_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_usd.csv"
cad_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_cad.csv"
cny_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_cny.csv"
eur_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_eur.csv"
bgp_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_bgp.csv"
jpy_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_jpy.csv"

# Load each dataset
df_usd = pd.read_csv(usd_path)
df_cad = pd.read_csv(cad_path)
df_cny = pd.read_csv(cny_path)
df_eur = pd.read_csv(eur_path)
df_bgp = pd.read_csv(bgp_path)
df_jpy = pd.read_csv(jpy_path)

# Drop any unwanted columns like 'Unnamed: 0'
df_usd = df_usd.drop(columns=[col for col in df_usd.columns if 'Unnamed' in col], errors='ignore')
df_cad = df_cad.drop(columns=[col for col in df_cad.columns if 'Unnamed' in col], errors='ignore')
df_cny = df_cny.drop(columns=[col for col in df_cny.columns if 'Unnamed' in col], errors='ignore')
df_eur = df_eur.drop(columns=[col for col in df_eur.columns if 'Unnamed' in col], errors='ignore')
df_bgp = df_bgp.drop(columns=[col for col in df_bgp.columns if 'Unnamed' in col], errors='ignore')
df_jpy = df_jpy.drop(columns=[col for col in df_jpy.columns if 'Unnamed' in col], errors='ignore')

# Ensure 'date' is in datetime format
df_usd['date'] = pd.to_datetime(df_usd['date'])
df_cad['date'] = pd.to_datetime(df_cad['date'])
df_cny['date'] = pd.to_datetime(df_cny['date'])
df_eur['date'] = pd.to_datetime(df_eur['date'])
df_bgp['date'] = pd.to_datetime(df_bgp['date'])
df_jpy['date'] = pd.to_datetime(df_jpy['date'])

# Merge the datasets on 'date'
df_all = df_usd.rename(columns={'rate': 'rate_usd'})
df_all = df_all.merge(df_cad.rename(columns={'rate': 'rate_cad'}), on='date')
df_all = df_all.merge(df_cny.rename(columns={'rate': 'rate_cny'}), on='date')
df_all = df_all.merge(df_eur.rename(columns={'rate': 'rate_eur'}), on='date')
df_all = df_all.merge(df_bgp.rename(columns={'rate': 'rate_bgp'}), on='date')
df_all = df_all.merge(df_jpy.rename(columns={'rate': 'rate_jpy'}), on='date')

# Now you have a DataFrame with all the interest rate data for each currency
print(df_all.head())

# Calculate the interest rate differentials for each currency pair
df_all['eur_usd'] = df_all['rate_eur'] - df_all['rate_usd']
df_all['gbp_usd'] = df_all['rate_bgp'] - df_all['rate_usd']
df_all['usd_jpy'] = df_all['rate_usd'] - df_all['rate_jpy']
df_all['usd_cad'] = df_all['rate_usd'] - df_all['rate_cad']
df_all['usd_cny'] = df_all['rate_usd'] - df_all['rate_cny']

# Check the resulting dataframe with interest rate differentials
print(df_all[['date', 'eur_usd', 'gbp_usd', 'usd_jpy', 'usd_cad', 'usd_cny']].head())
df_all = df_all.round(2)
output_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/merged_forex_interest_rates.csv"
df_all.to_csv(output_path, index=False)

print(f"Data saved to {output_path}")
```
