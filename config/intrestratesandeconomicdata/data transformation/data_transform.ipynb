```python
import pandas as pd

# Define file paths
usd_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_usd.csv"
cad_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_cad.csv"
cny_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_cny.csv"
eur_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_eur.csv"
bgp_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_bgp.csv"
jpy_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/processed_jpy.csv"

# Load each dataset
df_usd = pd.read_csv(usd_path)
df_cad = pd.read_csv(cad_path)
df_cny = pd.read_csv(cny_path)
df_eur = pd.read_csv(eur_path)
df_bgp = pd.read_csv(bgp_path)
df_jpy = pd.read_csv(jpy_path)

# Drop any unwanted columns like 'Unnamed: 0'
df_usd = df_usd.drop(columns=[col for col in df_usd.columns if 'Unnamed' in col], errors='ignore')
df_cad = df_cad.drop(columns=[col for col in df_cad.columns if 'Unnamed' in col], errors='ignore')
df_cny = df_cny.drop(columns=[col for col in df_cny.columns if 'Unnamed' in col], errors='ignore')
df_eur = df_eur.drop(columns=[col for col in df_eur.columns if 'Unnamed' in col], errors='ignore')
df_bgp = df_bgp.drop(columns=[col for col in df_bgp.columns if 'Unnamed' in col], errors='ignore')
df_jpy = df_jpy.drop(columns=[col for col in df_jpy.columns if 'Unnamed' in col], errors='ignore')

# Ensure 'date' is in datetime format
df_usd['date'] = pd.to_datetime(df_usd['date'])
df_cad['date'] = pd.to_datetime(df_cad['date'])
df_cny['date'] = pd.to_datetime(df_cny['date'])
df_eur['date'] = pd.to_datetime(df_eur['date'])
df_bgp['date'] = pd.to_datetime(df_bgp['date'])
df_jpy['date'] = pd.to_datetime(df_jpy['date'])

# Merge the datasets on 'date'
df_all = df_usd.rename(columns={'rate': 'rate_usd'})
df_all = df_all.merge(df_cad.rename(columns={'rate': 'rate_cad'}), on='date')
df_all = df_all.merge(df_cny.rename(columns={'rate': 'rate_cny'}), on='date')
df_all = df_all.merge(df_eur.rename(columns={'rate': 'rate_eur'}), on='date')
df_all = df_all.merge(df_bgp.rename(columns={'rate': 'rate_bgp'}), on='date')
df_all = df_all.merge(df_jpy.rename(columns={'rate': 'rate_jpy'}), on='date')

# Now you have a DataFrame with all the interest rate data for each currency
print(df_all.head())

# Calculate the interest rate differentials for each currency pair
df_all['eur_usd'] = df_all['rate_eur'] - df_all['rate_usd']
df_all['gbp_usd'] = df_all['rate_bgp'] - df_all['rate_usd']
df_all['usd_jpy'] = df_all['rate_usd'] - df_all['rate_jpy']
df_all['usd_cad'] = df_all['rate_usd'] - df_all['rate_cad']
df_all['usd_cny'] = df_all['rate_usd'] - df_all['rate_cny']

# Check the resulting dataframe with interest rate differentials
print(df_all[['date', 'eur_usd', 'gbp_usd', 'usd_jpy', 'usd_cad', 'usd_cny']].head())
df_all = df_all.round(2)
output_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/intrest_rates/merged_forex_interest_rates.csv"
df_all.to_csv(output_path, index=False)

print(f"Data saved to {output_path}")
```

Merging Inflation rates

```python
import pandas as pd
inflation_usd = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/CPI/usa_inflation_rate.csv", parse_dates=['date']).drop(columns=["unnamed: 0_x"], errors="ignore")
inflation_eur = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/CPI/euro_inflation_rate.csv", parse_dates=['date']).drop(columns=["unnamed: 0_x"], errors="ignore")
inflation_cad = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/CPI/cad_inflation_rate.csv", parse_dates=['date']).drop(columns=["unnamed: 0_x"], errors="ignore")
inflation_jpy = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/CPI/jpy_inflation_rate.csv", parse_dates=['date']).drop(columns=["unnamed: 0_x"], errors="ignore")
inflation_cny = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/CPI/cny_inflation_rate.csv", parse_dates=['date']).drop(columns=["unnamed: 0_x"], errors="ignore")
inflation_gbp = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/CPI/bgp_inflation_rate.csv", parse_dates=['date']).drop(columns=["unnamed: 0_x"], errors="ignore")

inflation_eur.rename(columns={"rate": "EUR"}, inplace=True)
inflation_usd.rename(columns={"rate": "USD"}, inplace=True)
inflation_gbp.rename(columns={"rate": "GBP"}, inplace=True)
inflation_cad.rename(columns={"rate": "CAD"}, inplace=True)
inflation_cny.rename(columns={"rate": "CNY"}, inplace=True)
inflation_jpy.rename(columns={"rate": "JPY"}, inplace=True)

# Merge datasets on the date column
merged = pd.merge(inflation_eur, inflation_usd, on="date", suffixes=("", "_USD"))
merged = pd.merge(merged, inflation_gbp, on="date", suffixes=("", "_GBP"))
merged = pd.merge(merged, inflation_cad, on="date", suffixes=("", "_CAD"))
merged = pd.merge(merged, inflation_cny, on="date", suffixes=("", "_CNY"))
merged = pd.merge(merged, inflation_jpy, on="date", suffixes=("", "_JPY"))

# Set date as index
merged.set_index("date", inplace=True)

# Calculate inflation differentials
differentials = pd.DataFrame(index=merged.index)
differentials["EUR/USD"] = merged["EUR"] - merged["USD"]
differentials["GBP/USD"] = merged["GBP"] - merged["USD"]
differentials["USD/CAD"] = merged["USD"] - merged["CAD"]
differentials["USD/CNY"] = merged["USD"] - merged["CNY"]
differentials["USD/JPY"] = merged["USD"] - merged["JPY"]

differentials.to_csv('C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/CPI/final_inflation_differentials_rate.csv')

print("Inflation differentials calculated and saved.")
```

FINAL GDP GROWTH RATE FILE

```python
gdp_usd = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/GDP/usa_gdp_growth_rate.csv", parse_dates=['date'])
gdp_eur = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/GDP/eur_gdp_growth_rate.csv", parse_dates=['date'])
gdp_cad = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/GDP/cad_gdp_growth_rate.csv", parse_dates=['date'])
gdp_jpy = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/GDP/jpy_gdp_growth_rate.csv", parse_dates=['date'])
gdp_cny = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/GDP/cny_gdp_growth_rate.csv", parse_dates=['date'])
gdp_gbp = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/GDP/gbp_gdp_growth_rate.csv", parse_dates=['date'])

# Drop unnecessary columns explicitly before renaming
gdp_usd = gdp_usd.drop(columns=["Unnamed: 0"], errors="ignore")
gdp_eur = gdp_eur.drop(columns=["Unnamed: 0"], errors="ignore")
gdp_cad = gdp_cad.drop(columns=["Unnamed: 0"], errors="ignore")
gdp_jpy = gdp_jpy.drop(columns=["Unnamed: 0"], errors="ignore")
gdp_cny = gdp_cny.drop(columns=["Unnamed: 0"], errors="ignore")
gdp_gbp = gdp_gbp.drop(columns=["Unnamed: 0"], errors="ignore")

# Rename columns
gdp_usd.rename(columns={"usa_gdp": "USD"}, inplace=True)
gdp_eur.rename(columns={"eur_gdp": "EUR"}, inplace=True)
gdp_cad.rename(columns={"cad_gdp": "CAD"}, inplace=True)
gdp_jpy.rename(columns={"jpy_gdp": "JPY"}, inplace=True)
gdp_cny.rename(columns={"cny_gdp": "CNY"}, inplace=True)
gdp_gbp.rename(columns={"gbp_gdp": "GBP"}, inplace=True)

# Merge dataframes
gdp_data = gdp_usd.merge(gdp_eur, on="date", suffixes=('_usd', '_eur')) \
                  .merge(gdp_gbp, on="date", suffixes=('_eur', '_gbp')) \
                  .merge(gdp_jpy, on="date", suffixes=('_gbp', '_jpy')) \
                  .merge(gdp_cny, on="date", suffixes=('_jpy', '_cny')) \
                  .merge(gdp_cad, on="date", suffixes=('_cny', '_cad'))

# Add currency pair columns
currency_pairs = {
    "EUR/USD": ("EUR", "USD"),
    "GBP/USD": ("GBP", "USD"),
    "USD/JPY": ("USD", "JPY"),
    "USD/CNY": ("USD", "CNY"),
    "USD/CAD": ("USD", "CAD")
}

for pair, (base, quote) in currency_pairs.items():
    gdp_data[pair] = gdp_data[base] - gdp_data[quote]

# Save to CSV
gdp_data.to_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/GDP/final_gdp_growth_rate.csv", index=False)
print(gdp_data.head())
```

```python
import pandas as pd

unemployment_usd = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/Unemployment rate/usa_unemployment_rate.csv", parse_dates=['date'])
unemployment_eur = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/Unemployment rate/eur_unemployment_rate.csv", parse_dates=['date'])
unemployment_cad = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/Unemployment rate/cad_unemployment_rate.csv", parse_dates=['date'])
unemployment_jpy = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/Unemployment rate/jpy_unemployment_rate.csv", parse_dates=['date'])
unemployment_cny = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/Unemployment rate/cny_unemployment_rate.csv", parse_dates=['date'])
unemployment_gbp = pd.read_csv("C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/Unemployment rate/gbp_unemployment_rate.csv", parse_dates=['date'])


unemployment_usd = unemployment_usd.drop(columns=["Unnamed: 0"], errors="ignore")
unemployment_eur = unemployment_eur.drop(columns=["Unnamed: 0"], errors="ignore")
unemployment_cad = unemployment_cad.drop(columns=["Unnamed: 0"], errors="ignore")
unemployment_jpy = unemployment_jpy.drop(columns=["Unnamed: 0"], errors="ignore")
unemployment_cny = unemployment_cny.drop(columns=["Unnamed: 0"], errors="ignore")
unemployment_gbp = unemployment_gbp.drop(columns=["Unnamed: 0"], errors="ignore")

unemployment_usd.rename(columns={"USA Unemployment rate": "USD Unemployment Rate"}, inplace=True)
unemployment_eur.rename(columns={"EUR Unemployment rate": "EUR Unemployment Rate"}, inplace=True)
unemployment_cad.rename(columns={"CAD Unemployment rate": "CAD Unemployment Rate"}, inplace=True)
unemployment_jpy.rename(columns={"JPY Unemployment rate": "JPY Unemployment Rate"}, inplace=True)
unemployment_cny.rename(columns={"CNY Unemployment rate": "CNY Unemployment Rate"}, inplace=True)
unemployment_gbp.rename(columns={"GBP Unemployment rate": "GBP Unemployment Rate"}, inplace=True)

merged_df = unemployment_usd
dataframes = [unemployment_eur, unemployment_cad, unemployment_jpy, unemployment_cny, unemployment_gbp]


for df in dataframes:
    merged_df = pd.merge(merged_df, df, on='date', how='inner') 

currency_columns = [col for col in merged_df.columns if col != 'date']


for i in range(len(currency_columns)):
    for j in range(i + 1, len(currency_columns)):
        currency_1 = currency_columns[i]
        currency_2 = currency_columns[j]
        pair_ratio = f"{currency_1.split()[0]}/{currency_2.split()[0]} Unemployment Ratio"
        pair_diff = f"{currency_1.split()[0]}/{currency_2.split()[0]} Unemployment Diff"
        merged_df[pair_ratio] = merged_df[currency_1] / merged_df[currency_2]
        merged_df[pair_diff] = merged_df[currency_1] - merged_df[currency_2]


merged_df.index = merged_df.index + 1
output_path = "C:/Users/vamsh/OneDrive/Desktop/Forex trading/Forex-Trading/config/intrestratesandeconomicdata/data/Unemployment rate/final_unemployment_rate.csv"
merged_df.to_csv(output_path, index=True)

print("Processed unemployment data saved successfully.")
print(merged_df.head())
```
